```python
"""
Extract all unique routes from KSRTC data and create route optimization data
This script reads KSRTC.csv and generates route analysis for all routes
"""

import pandas as pd
import json
from collections import Counter

# Read KSRTC data
df = pd.read_csv('D:\\Rachith Bharadwaj T N 24BDS062\\2nd Year\\3rd SEM\\AI Project\\KSRTC_Project_Submission\\3.Frontend\\2.commuter-app\\app\\public\\ksrtc.csv')

# Get all unique routes
routes = df['route'].unique() if 'route' in df.columns else []

print(f"✅ Found {len(routes)} unique routes in KSRTC.csv")
print("\n" + "="*80)
print("ALL ROUTES IN YOUR DATA:")
print("="*80)

routes_list = []
for i, route in enumerate(routes, 1):
    print(f"{i}. {route}")
    routes_list.append(route)

# Save routes to JSON for reference
routes_data = {
    "total_routes": len(routes),
    "routes": routes_list
}

with open('routes_list.json', 'w') as f:
    json.dump(routes_data, f, indent=2)

print("\n" + "="*80)
print(f"✅ Saved {len(routes)} routes to 'routes_list.json'")
print("="*80)

# Display sample route data
print("\nSample data for each route:")
print("-" * 80)

for route in routes[:5]:
    route_data = df[df['route'] == route]
    print(f"\n📍 Route: {route}")
    print(f"   - Total records: {len(route_data)}")
    print(f"   - Columns available: {list(df.columns)}")
    print(f"   - Sample: {route_data.head(1).to_dict()}")
```
