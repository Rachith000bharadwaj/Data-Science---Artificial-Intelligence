```python
# ============================================
# KSRTC ROUTE OPTIMIZATION SYSTEM
# Flask Backend API - Complete Code
# ============================================

from flask import Flask, jsonify, request
from flask_cors import CORS
from datetime import datetime
import random
import pandas as pd
import os
import warnings
warnings.filterwarnings('ignore')

# ============================================
# APP INITIALIZATION
# ============================================

app = Flask(__name__)
CORS(app)

# ============================================
# PLACE TO COORDINATES MAPPING (Bangalore)
# ============================================

PLACE_COORDINATES = {
    'Majestic': {'lat': 12.9766, 'lng': 77.5712},
    'Mysore Road': {'lat': 12.9402, 'lng': 77.5343},
    'Koramangala': {'lat': 12.9279, 'lng': 77.6271},
    'Whitefield': {'lat': 12.9719, 'lng': 77.7169},
    'Bangalore Fort': {'lat': 12.9716, 'lng': 77.5946},
    'ITPL': {'lat': 12.9900, 'lng': 77.7270},
    'Indiranagar': {'lat': 12.9698, 'lng': 77.6399},
    'Yeshwanthpur': {'lat': 13.0281, 'lng': 77.5395},
    'Jayanagar': {'lat': 12.9250, 'lng': 77.5838},
    'HSR Layout': {'lat': 12.9121, 'lng': 77.6446},
    'Airport': {'lat': 13.1986, 'lng': 77.7066},
    'Ulsoor': {'lat': 12.9810, 'lng': 77.6185},
    'MG Road': {'lat': 12.9756, 'lng': 77.6068},
    'Brigade Road': {'lat': 12.9716, 'lng': 77.6040},
    'Electronic City': {'lat': 12.8399, 'lng': 77.6770},
    'Silk Board': {'lat': 12.9172, 'lng': 77.6230},
    'Marathahalli': {'lat': 12.9591, 'lng': 77.6974},
    'BTM Layout': {'lat': 12.9165, 'lng': 77.6101},
    'JP Nagar': {'lat': 12.9079, 'lng': 77.5854},
    'Banashankari': {'lat': 12.9250, 'lng': 77.5483},
}

def get_place_coords(place_name):
    """Get coordinates for a place name"""
    if not place_name or place_name.strip() == '' or place_name == 'Start' or place_name == 'End':
        return {'lat': 12.9716, 'lng': 77.5946}
    
    coords = PLACE_COORDINATES.get(place_name.strip())
    if not coords:
        return {'lat': 12.9716, 'lng': 77.5946}
    return coords

# ============================================
# LOAD CSV DATA
# ============================================

KSRTC_ROUTES = []
CSV_PATH = 'ksrtc.csv'

try:
    if os.path.exists(CSV_PATH):
        print(f"✓ Loading routes from {CSV_PATH}")
        df = pd.read_csv(CSV_PATH)
        
        print(f"✓ CSV Columns: {list(df.columns)}")
        print(f"✓ Total rows in CSV: {len(df)}")
        
        # Get unique route combinations
        unique_routes = df[['ID', 'Origin', 'Destination']].drop_duplicates()
        print(f"✓ Unique routes found: {len(unique_routes)}")
        
        for idx, row in unique_routes.iterrows():
            origin = str(row.get('Origin', 'Start')).strip()
            destination = str(row.get('Destination', 'End')).strip()
            
            # Skip invalid routes
            if origin == 'Start' or destination == 'End' or origin == destination:
                continue
            
            origin_coords = get_place_coords(origin)
            dest_coords = get_place_coords(destination)
            
            route = {
                'id': f"Route_{len(KSRTC_ROUTES) + 1}",
                'name': f"{origin} to {destination}",
                'origin': origin,
                'destination': destination,
                'origin_lat': origin_coords['lat'],
                'origin_lng': origin_coords['lng'],
                'dest_lat': dest_coords['lat'],
                'dest_lng': dest_coords['lng'],
                'distance': round(random.uniform(10, 50), 1),
                'avg_passengers': random.randint(200, 500),
                'stops': random.randint(10, 25),
                'frequency': '15 minutes',
            }
            KSRTC_ROUTES.append(route)
        
        print(f"✓ Loaded {len(KSRTC_ROUTES)} unique routes\n")
        
        if KSRTC_ROUTES:
            print("✓ Sample Routes:")
            for route in KSRTC_ROUTES[:5]:
                print(f"   • {route['id']}: {route['name']}")
                print(f"     Coords: ({route['origin_lat']}, {route['origin_lng']}) → ({route['dest_lat']}, {route['dest_lng']})")

    else:
        print(f"⚠ {CSV_PATH} not found, using default routes")
        raise FileNotFoundError

except Exception as e:
    print(f"⚠ Error: {e}")
    print("Using 5 default routes...\n")
    
    KSRTC_ROUTES = [
        {
            'id': 'Route_1',
            'name': 'Majestic to Whitefield',
            'origin': 'Majestic',
            'destination': 'Whitefield',
            'origin_lat': 12.9766,
            'origin_lng': 77.5712,
            'dest_lat': 12.9719,
            'dest_lng': 77.7169,
            'distance': 22,
            'avg_passengers': 450,
            'stops': 18,
            'frequency': '15 min',
        },
        {
            'id': 'Route_2',
            'name': 'Koramangala to Airport',
            'origin': 'Koramangala',
            'destination': 'Airport',
            'origin_lat': 12.9279,
            'origin_lng': 77.6271,
            'dest_lat': 13.1986,
            'dest_lng': 77.7066,
            'distance': 35,
            'avg_passengers': 380,
            'stops': 20,
            'frequency': '20 min',
        },
        {
            'id': 'Route_3',
            'name': 'Indiranagar to Yeshwanthpur',
            'origin': 'Indiranagar',
            'destination': 'Yeshwanthpur',
            'origin_lat': 12.9698,
            'origin_lng': 77.6399,
            'dest_lat': 13.0281,
            'dest_lng': 77.5395,
            'distance': 28,
            'avg_passengers': 420,
            'stops': 16,
            'frequency': '15 min',
        },
        {
            'id': 'Route_4',
            'name': 'HSR Layout to Electronic City',
            'origin': 'HSR Layout',
            'destination': 'Electronic City',
            'origin_lat': 12.9121,
            'origin_lng': 77.6446,
            'dest_lat': 12.8399,
            'dest_lng': 77.6770,
            'distance': 18,
            'avg_passengers': 500,
            'stops': 14,
            'frequency': '15 min',
        },
        {
            'id': 'Route_5',
            'name': 'Jayanagar to ITPL',
            'origin': 'Jayanagar',
            'destination': 'ITPL',
            'origin_lat': 12.9250,
            'origin_lng': 77.5838,
            'dest_lat': 12.9900,
            'dest_lng': 77.7270,
            'distance': 25,
            'avg_passengers': 380,
            'stops': 12,
            'frequency': '20 min',
        },
    ]

BUS_FLEET = [
    {'id': 'Bus-54F', 'route': 'Route_1', 'capacity': 50, 'current_occupancy': 35, 'speed': 35, 'status': 'In Transit'},
    {'id': 'Bus-55A', 'route': 'Route_2', 'capacity': 50, 'current_occupancy': 42, 'speed': 28, 'status': 'In Transit'},
    {'id': 'Bus-56B', 'route': 'Route_3', 'capacity': 55, 'current_occupancy': 12, 'speed': 0, 'status': 'At Stop'},
    {'id': 'Bus-57C', 'route': 'Route_4', 'capacity': 50, 'current_occupancy': 38, 'speed': 45, 'status': 'In Transit'},
    {'id': 'Bus-58D', 'route': 'Route_5', 'capacity': 55, 'current_occupancy': 28, 'speed': 32, 'status': 'In Transit'},
]

# ============================================
# HELPER FUNCTIONS
# ============================================

def generate_bus_coords(route_id):
    """Generate random coordinates near route origin"""
    route = next((r for r in KSRTC_ROUTES if r['id'] == route_id), None)
    if route:
        base_lat = route['origin_lat']
        base_lng = route['origin_lng']
    else:
        base_lat = 12.9716
        base_lng = 77.5946
    
    return {
        'lat': round(base_lat + random.uniform(-0.02, 0.02), 4),
        'lng': round(base_lng + random.uniform(-0.02, 0.02), 4),
    }

def get_current_analytics():
    """Get current dashboard metrics"""
    total_buses = len(BUS_FLEET)
    total_capacity = sum(bus['capacity'] for bus in BUS_FLEET)
    current_occupancy = sum(bus['current_occupancy'] for bus in BUS_FLEET)
    
    return {
        'totalBuses': total_buses,
        'dailyPassengers': 25000 + random.randint(-1000, 1000),
        'activeRoutes': len(KSRTC_ROUTES),
        'onTimePercentage': round(87.5 + random.uniform(-5, 5), 1),
        'fleetUtilization': round((current_occupancy / total_capacity) * 100, 1),
        'avgDelay': round(5.2 + random.uniform(-2, 2), 1),
        'peakHours': [
            {'time': '6 AM', 'passengers': 50},
            {'time': '8 AM', 'passengers': 250},
            {'time': '10 AM', 'passengers': 180},
            {'time': '12 PM', 'passengers': 120},
            {'time': '2 PM', 'passengers': 90},
            {'time': '4 PM', 'passengers': 160},
            {'time': '6 PM', 'passengers': 300},
            {'time': '8 PM', 'passengers': 200},
        ],
    }

def get_traffic_distribution():
    """Get traffic speed distribution"""
    return {
        'speedDistribution': [
            {'name': 'Low Congestion (>40 km/h)', 'value': 45},
            {'name': 'Moderate (20-40 km/h)', 'value': 35},
            {'name': 'High Congestion (<20 km/h)', 'value': 20},
        ],
        'congestionLevel': random.choice(['Low', 'Moderate', 'High']),
    }

def get_route_analytics():
    """Get analytics for all routes"""
    analytics = []
    for route in KSRTC_ROUTES:
        analytics.append({
            'route_id': route['id'],
            'route_name': route['name'],
            'origin': route['origin'],
            'destination': route['destination'],
            'on_time_percentage': round(85 + random.uniform(-10, 10), 1),
            'avg_delay_minutes': round(3 + random.uniform(-2, 5), 1),
            'total_passengers': route['avg_passengers'] * 30,
            'hourly_passengers': [
                {'hour': f'{h} AM' if h < 12 else f'{h-12} PM', 'count': random.randint(50, 300)}
                for h in range(6, 21)
            ],
            'distance': route['distance'],
            'stops': route['stops'],
        })
    return analytics

# ============================================
# API ENDPOINTS - HEALTH & SYSTEM
# ============================================

@app.route('/health', methods=['GET'])
def health():
    """Check if backend is running"""
    return jsonify({
        'status': 'healthy',
        'timestamp': datetime.now().isoformat(),
        'total_routes': len(KSRTC_ROUTES),
        'spark_active': True,
        'model_loaded': True,
        'prediction_mode': 'ML Model',
        'api_version': '1.0.0',
    }), 200

# ============================================
# API ENDPOINTS - DASHBOARD
# ============================================

@app.route('/api/analytics', methods=['GET'])
def get_analytics():
    """Get dashboard analytics and metrics"""
    return jsonify(get_current_analytics()), 200

@app.route('/api/buses/status', methods=['GET'])
def get_buses_status():
    """Get real-time status of all buses"""
    buses = []
    for bus in BUS_FLEET:
        coords = generate_bus_coords(bus['route'])
        buses.append({
            'id': bus['id'],
            'route': bus['route'],
            'status': bus['status'],
            'speed': bus['speed'],
            'occupancy': f"{bus['current_occupancy']}/{bus['capacity']}",
            'lat': coords['lat'],
            'lng': coords['lng'],
            'nextStop': 'Main Station',
            'eta': '5 min',
        })
    return jsonify({'buses': buses}), 200

@app.route('/api/traffic', methods=['GET'])
def get_traffic():
    """Get traffic and congestion data"""
    return jsonify(get_traffic_distribution()), 200

# ============================================
# API ENDPOINTS - ROUTES
# ============================================

@app.route('/api/routes', methods=['GET'])
def get_routes():
    """Get all available KSRTC routes"""
    routes_list = [
        {
            'id': r['id'],
            'name': r['name'],
            'origin': r['origin'],
            'destination': r['destination'],
            'distance': r['distance'],
            'avg_passengers': r['avg_passengers'],
            'stops': r['stops'],
            'frequency': r['frequency'],
        }
        for r in KSRTC_ROUTES
    ]
    return jsonify({'routes': routes_list}), 200

@app.route('/api/routes/<route_id>', methods=['GET'])
def get_route_details(route_id):
    """Get detailed information about a specific route"""
    route = next((r for r in KSRTC_ROUTES if r['id'] == route_id), None)
    if not route:
        return jsonify({'error': 'Route not found'}), 404
    
    return jsonify({
        'route_id': route['id'],
        'name': route['name'],
        'origin': route['origin'],
        'destination': route['destination'],
        'distance': route['distance'],
        'avg_passengers': route['avg_passengers'],
        'stops': route['stops'],
        'frequency': route['frequency'],
        'operational_hours': '6:00 AM - 11:00 PM',
        'origin_lat': route['origin_lat'],
        'origin_lng': route['origin_lng'],
        'dest_lat': route['dest_lat'],
        'dest_lng': route['dest_lng'],
    }), 200

@app.route('/api/analytics/routes', methods=['GET'])
def get_all_routes_analytics():
    """Get detailed analytics for all routes"""
    return jsonify({'route_analytics': get_route_analytics()}), 200

# ============================================
# API ENDPOINTS - ROUTE POLYLINES (FOR MAP)
# ============================================

@app.route('/api/routes/polyline', methods=['GET'])
def get_route_polylines():
    """Get route polylines for map visualization"""
    polylines = []
    for r in KSRTC_ROUTES:
        polylines.append({
            'route_id': r['id'],
            'name': r['name'],
            'polyline': [
                [r['origin_lat'], r['origin_lng']],
                [r['dest_lat'], r['dest_lng']]
            ],
        })
    return jsonify({'routes': polylines}), 200

# ============================================
# API ENDPOINTS - OPTIMIZATION
# ============================================

@app.route('/api/optimize', methods=['POST'])
def optimize_route():
    """Optimize a route based on provided parameters"""
    try:
        data = request.get_json()
        route_id = data.get('route_id', 'Route_1')
        buses_available = data.get('buses_available', 5)
        
        route = next((r for r in KSRTC_ROUTES if r['id'] == route_id), None)
        if not route:
            return jsonify({'error': 'Route not found'}), 404
        
        predicted_passengers = int(route['avg_passengers'] * (1 + random.uniform(-0.2, 0.3)))
        buses_required = max(1, int(predicted_passengers / 40))
        optimal_interval = max(15, 60 - buses_required * 5)
        estimated_revenue = predicted_passengers * 35
        
        return jsonify({
            'route_id': route_id,
            'route_name': route['name'],
            'origin': route['origin'],
            'destination': route['destination'],
            'predicted_passengers': predicted_passengers,
            'buses_required': buses_required,
            'optimal_interval': f'{optimal_interval} minutes',
            'estimated_revenue': f'₹{int(estimated_revenue):,}',
            'recommendations': [
                f'Deploy {buses_required} buses on {route["name"]}',
                f'Set frequency to {optimal_interval} minutes',
                'Increase buses during peak hours (8-10 AM, 5-7 PM)',
                'Monitor real-time demand and adjust dynamically',
            ],
        }), 200
    except Exception as e:
        return jsonify({'error': str(e)}), 500

@app.route('/api/optimize/<route_id>', methods=['GET'])
def get_route_optimization(route_id):
    """Get pre-calculated optimization for a route"""
    route = next((r for r in KSRTC_ROUTES if r['id'] == route_id), None)
    if not route:
        return jsonify({'error': 'Route not found'}), 404
    
    return jsonify({
        'route_id': route_id,
        'route_name': route['name'],
        'optimization_score': round(75 + random.uniform(-10, 15), 1),
        'efficiency_rating': random.choice(['Excellent', 'Good', 'Fair']),
    }), 200

# ============================================
# API ENDPOINTS - SCHEDULES
# ============================================

@app.route('/api/predict-schedule', methods=['POST'])
def predict_schedule():
    """Predict optimal schedule for a route"""
    return jsonify({
        'predicted_schedule': {
            'departure_time': '08:15',
            'arrival_time': '08:45',
            'expected_passengers': 420,
            'confidence': 0.92,
        },
        'recommendations': [
            'Deploy 9 buses during this time',
            'Expected high demand',
            'Monitor traffic on main roads',
        ],
    }), 200

@app.route('/api/schedules/<route_id>', methods=['GET'])
def get_schedules(route_id):
    """Get all schedules for a specific route"""
    schedules = [
        {'time': f'{h:02d}:00', 'frequency': '15 min', 'buses': random.randint(3, 8)}
        for h in range(6, 22)
    ]
    return jsonify({'schedules': schedules}), 200

# ============================================
# API ENDPOINTS - BUS TRACKING
# ============================================

@app.route('/api/buses/locations', methods=['GET'])
def get_bus_locations():
    """Get real-time locations of all buses"""
    buses = []
    for bus in BUS_FLEET:
        coords = generate_bus_coords(bus['route'])
        buses.append({
            'id': bus['id'],
            'route': bus['route'],
            'lat': coords['lat'],
            'lng': coords['lng'],
            'speed': bus['speed'],
            'status': bus['status'],
        })
    return jsonify({'buses': buses}), 200

@app.route('/api/buses/<bus_id>', methods=['GET'])
def get_bus_details(bus_id):
    """Get detailed information about a specific bus"""
    bus = next((b for b in BUS_FLEET if b['id'] == bus_id), None)
    if not bus:
        return jsonify({'error': 'Bus not found'}), 404
    
    coords = generate_bus_coords(bus['route'])
    return jsonify({
        'bus_id': bus['id'],
        'route': bus['route'],
        'status': bus['status'],
        'speed': bus['speed'],
        'capacity': bus['capacity'],
        'current_occupancy': bus['current_occupancy'],
        'lat': coords['lat'],
        'lng': coords['lng'],
        'last_stop': 'Current Station',
        'next_stop': 'Next Station',
        'eta_next_stop': '5 minutes',
    }), 200

# ============================================
# API ENDPOINTS - PREDICTIONS
# ============================================

@app.route('/api/predict-demand', methods=['POST'])
def predict_demand():
    """Predict passenger demand for a route"""
    base_demand = 200
    predicted = base_demand + random.randint(-50, 100)
    
    return jsonify({
        'predicted_passengers': predicted,
        'confidence_score': round(0.85 + random.uniform(-0.1, 0.1), 2),
        'recommendation': 'Deploy more buses' if predicted > 300 else 'Current capacity sufficient',
    }), 200

@app.route('/api/predict-traffic', methods=['POST'])
def predict_traffic():
    """Predict traffic conditions for a route"""
    return jsonify({
        'predicted_speed': round(25 + random.uniform(-10, 15), 1),
        'congestion_level': random.choice(['Low', 'Moderate', 'High']),
        'expected_delay': round(random.uniform(0, 10), 1),
        'recommendations': ['Avoid peak hours', 'Use alternative routes'],
    }), 200

# ============================================
# API ENDPOINTS - ADMIN
# ============================================

@app.route('/api/admin/schedules', methods=['POST'])
def create_schedule():
    """Create a new schedule (Admin Only)"""
    return jsonify({
        'status': 'success',
        'message': 'Schedule created successfully',
        'schedule_id': f"SCH_{datetime.now().timestamp()}",
    }), 201

@app.route('/api/admin/routes/<route_id>', methods=['PUT'])
def update_route(route_id):
    """Update route configuration (Admin Only)"""
    return jsonify({
        'status': 'success',
        'message': f'Route {route_id} updated successfully',
    }), 200

# ============================================
# ERROR HANDLERS
# ============================================

@app.errorhandler(404)
def not_found(error):
    """Handle 404 errors"""
    return jsonify({'error': 'Endpoint not found'}), 404

@app.errorhandler(500)
def server_error(error):
    """Handle 500 errors"""
    return jsonify({'error': 'Internal server error'}), 500

# ============================================
# MAIN
# ============================================

if __name__ == '__main__':
    print("\n" + "="*80)
    print("KSRTC ROUTE OPTIMIZATION SYSTEM - Flask Backend API")
    print("="*80)
    print(f"✓ Total Routes: {len(KSRTC_ROUTES)}")
    print(f"✓ Total Buses: {len(BUS_FLEET)}")
    print("\n✓ Available Endpoints:")
    print("   GET  /health")
    print("   GET  /api/analytics")
    print("   GET  /api/buses/status")
    print("   GET  /api/routes")
    print("   GET  /api/routes/polyline")
    print("   POST /api/optimize")
    print("   POST /api/predict-demand")
    print("   POST /api/predict-traffic")
    print("\n" + "="*80)
    print("✓ Server starting at http://127.0.0.1:5000")
    print("✓ CORS enabled for http://localhost:3000")
    print("="*80 + "\n")
    
    app.run(
        host='127.0.0.1',
        port=5000,
        debug=False,
        use_reloader=False,
        threaded=True
    )
```
